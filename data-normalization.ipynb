# Data Normalization: Rescaling

```python
# Config
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import RobustScaler

data = pd.read_csv("...") # Data Source/Path
```

```python
# Show dataset

data.head(11)
```

```python
# Show DataTypes

data.dtypes
```

**1. Rescaling: Min-Max Normalization**

```python
# Rescaling: MinMaxScaler

cols = ['Engine HP', 'Engine Cylinders', 'highway MPG', 'city mpg', 'MSRP']

scaler_minmax = MinMaxScaler()
scaled_minmax = scaler_minmax.fit_transform(data[cols])

scaled_minmax_df = pd.DataFrame(scaled_minmax, columns=[f'{col}_scaled' for col in cols])
pd.concat([data[cols], scaled_minmax_df], axis=1).head(11)
```

**2. Rescaling: RobustScaler**

```python
scaler_robust = RobustScaler()
scaled_robust = scaler_robust.fit_transform(data[cols])

scaled_robust_df = pd.DataFrame(scaled_robust, columns=[f'{col}_scaled' for col in cols])
pd.concat([data[cols], scaled_robust_df], axis=1).head(11)
```

**3. Comparison**

```python
comparison = pd.DataFrame({
    'Columns': cols,
    'Min (Original)': data[cols].min().values,
    'Max (Original)': data[cols].max().values,
    'Min (MinMax)': scaled_minmax_df.min().values,
    'Max (MinMax)': scaled_minmax_df.max().values,
    'Min (Robust)': scaled_robust_df.min().values,
    'Max (Robust)': scaled_robust_df.max().values,
})

comparison
```
