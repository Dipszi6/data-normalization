## Data Normalization: Mean Normalization

```python
import pandas as pd

data = pd.read_csv("...") # File Path
```

```python
data.dtypes
```

```python
data = data.drop(['Engine Fuel Type', 'Engine Cylinders', 'Driven_Wheels', 'Number of Doors', 'Market Category', 'Vehicle Size', 'Vehicle Style'], axis=1)

data.head(10)
```

**1. Mean Normalization**

```python
cols = ['Engine HP', 'highway MPG', 'city mpg', 'MSRP']

mean_scaled = (data[cols] - data[cols].mean()) / (data[cols].max() - data[cols].min())

mean_scaled_df = mean_scaled.add_suffix('_scaled')

comparasion_mean = pd.concat([data[cols], mean_scaled_df], axis=1)
comparasion_mean.head(11)
```

**2. Comparasion**

```python
comparison = pd.DataFrame({
    'Kolom': cols,
    'Min (Original)': data[cols].min().values,
    'Max (Original)': data[cols].max().values,
    'Min (Mean Norm)': mean_scaled_df.min().values,
    'Max (Mean Norm)': mean_scaled_df.max().values,
    })

comparison
```
